```python
# openpyxl: Excelファイルを作成・編集できるライブラリ
import openpyxl

# pandas: 表形式データ（テーブルデータ）を扱うライブラリ
import pandas as pd

from openpyxl.styles import PatternFill

# 処理対象のファイル名のリストを格納
file_list = ['/content/2022_年間売上表.xlsx', '/content/2023_年間売上表.xlsx']

# 結合後のデータを格納するための空のデータフレームを作成
df = pd.DataFrame()

# ファイルを一つずつ処理する
for file_name in file_list:
    # Excelファイルを読み込む
    temp_df = pd.read_excel(file_name)

    # データフレームの列名を指定
    temp_df.columns = ['売上年', '商品', '金額']

    # concat()でデータフレームを連結
    df = pd.concat([df, temp_df], ignore_index=True)

print(df)

# 商品、売上年ごとにグループ化、「売上金額」列の合計を計算する
result = df.groupby(['売上年', '商品'])['金額'].sum()

print(result)

# resultを並び替え
result2 = result.sort_index(level='商品')
result2_df = result2.reset_index()
print("rusult2----------")
print(result2_df)

# Excelファイルを作成
writer = pd.ExcelWriter('売上集計表.xlsx')

# DataFrameオブジェクトをExcelファイルに書き込む
result2_df.to_excel(writer, sheet_name='Sheet1', index=False)


# Excelファイルを保存する
writer.close()

# ワークブックを作成
workbook = openpyxl.load_workbook('売上集計表.xlsx')

# アクティブなシートを変数「sheet」に格納
sheet = workbook.active



# 1行目の背景色を薄いグレー色（#FF0000）に設定
gray_fill = PatternFill(start_color="F2F2F2", end_color="F2F2F2", fill_type="solid")

for cell in sheet[1]:
    cell.fill = gray_fill

workbook.save('売上集計表.xlsx')
```
